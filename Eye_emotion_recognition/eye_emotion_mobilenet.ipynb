The code for emotion detection using full face data 

```python
import os
import cv2
import numpy as np
from collections import Counter
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.utils import class_weight
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications.mobilenet_v2 import MobileNetV2, preprocess_input
from tensorflow.keras import layers, models, optimizers, callbacks

# Constants
IMG_SIZE = (224, 224)
BATCH_SIZE = 32
EPOCHS = 50
CLASS_NAMES = ['angry', 'fear', 'happy', 'sad']
DATA_DIR = r"D:\Dataset_2000_128_128\4_classes"  # Change to your full face image dataset path

# Load OpenCV's Haar cascade face detector
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_frontalface_default.xml")

def normalize_brightness_contrast(img):
    lab = cv2.cvtColor(img, cv2.COLOR_RGB2LAB)
    l, a, b = cv2.split(lab)
    clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
    cl = clahe.apply(l)
    limg = cv2.merge((cl, a, b))
    return cv2.cvtColor(limg, cv2.COLOR_LAB2RGB)

def detect_and_crop_face(image):
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    faces = face_cascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=5, minSize=(100, 100))
    if len(faces) == 0:
        return None  # No face detected
    # Assuming largest detected face is the main face
    largest_face = max(faces, key=lambda rect: rect[2] * rect[3])
    x, y, w, h = largest_face
    face_img = image[y:y+h, x:x+w]
    return face_img

def load_and_preprocess_all(data_dir):
    images, labels = [], []
    class_counts = {cls: 0 for cls in CLASS_NAMES}
    for idx, cls in enumerate(CLASS_NAMES):
        cls_path = os.path.join(data_dir, cls)
        if not os.path.exists(cls_path):
            print(f"Warning: class folder {cls} missing.")
            continue
        files = [f for f in os.listdir(cls_path) if os.path.isfile(os.path.join(cls_path, f))]
        class_counts[cls] = len(files)
        print(f"Loading {class_counts[cls]} '{cls}' images")
        for file in files:
            img_path = os.path.join(cls_path, file)
            img = cv2.imread(img_path)
            if img is None:
                continue
            face = detect_and_crop_face(img)
            if face is None:
                continue  # Skip images where no face detected
            img_resized = cv2.resize(face, IMG_SIZE)
            img_rgb = cv2.cvtColor(img_resized, cv2.COLOR_BGR2RGB)
            img_norm = normalize_brightness_contrast(img_rgb)
            preprocessed = preprocess_input(img_norm)
            images.append(preprocessed)
            labels.append(idx)
    images = np.array(images)
    labels = np.array(labels)
    print(f"Loaded {len(images)} images with class counts: {class_counts}")
    return images, labels

def build_model(input_shape=(*IMG_SIZE, 3), num_classes=len(CLASS_NAMES)):
    base_model = MobileNetV2(include_top=False, input_shape=input_shape, pooling='avg', weights='imagenet')
    base_model.trainable = True
    for layer in base_model.layers[:-20]:
        layer.trainable = False
    model = models.Sequential([
        base_model,
        layers.BatchNormalization(),
        layers.Dense(128, activation='relu'),
        layers.Dropout(0.5),
        layers.Dense(num_classes, activation='softmax')
    ])
    model.compile(optimizer=optimizers.Adam(1e-4),
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])
    return model

def plot_training_history(history):
    plt.figure(figsize=(12, 5))
    plt.subplot(1, 2, 1)
    plt.plot(history.history['accuracy'], label='Train Accuracy')
    plt.plot(history.history.get('val_accuracy', []), label='Val Accuracy')
    plt.title('Accuracy')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy')
    plt.legend()
    plt.subplot(1, 2, 2)
    plt.plot(history.history['loss'], label='Train Loss')
    plt.plot(history.history.get('val_loss', []), label='Val Loss')
    plt.title('Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    plt.tight_layout()
    plt.show()

# Load dataset
X, y = load_and_preprocess_all(DATA_DIR)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

# Calculate class weights for imbalanced classes
class_weights = class_weight.compute_class_weight(class_weight='balanced',
                                                  classes=np.unique(y_train),
                                                  y=y_train)
class_weights_dict = dict(enumerate(class_weights))
print("Class weights:", class_weights_dict)

# Data augmentation for training
train_datagen = ImageDataGenerator(rotation_range=30, shear_range=0.3, zoom_range=0.3,
                                   width_shift_range=0.15, height_shift_range=0.15,
                                   brightness_range=[0.7, 1.3],
                                   horizontal_flip=True, fill_mode='nearest')
val_datagen = ImageDataGenerator()

# Build and train model
model = build_model()
model.summary()

early_stopping = callbacks.EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True, mode='min')
reduce_lr = callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=5, min_lr=1e-6, mode='min')

history = model.fit(train_datagen.flow(X_train, y_train, batch_size=BATCH_SIZE),
                    validation_data=val_datagen.flow(X_test, y_test, batch_size=BATCH_SIZE),
                    epochs=EPOCHS,
                    class_weight=class_weights_dict,
                    callbacks=[early_stopping, reduce_lr])

plot_training_history(history)

# Save the trained model to a file
model.save("emotion_model_full_face_mobileNet.keras")
print("Model saved to emotion_model_full_face.keras")
```

Trained Gray Scale Images

```python
import os
import cv2
import numpy as np
from collections import Counter
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.utils import class_weight
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications.mobilenet_v2 import MobileNetV2, preprocess_input
from tensorflow.keras import layers, models, optimizers, callbacks

# Constants
IMG_SIZE = (224, 224)
BATCH_SIZE = 32
EPOCHS = 50
CLASS_NAMES = ['angry', 'fear', 'happy', 'sad']
DATA_DIR = r"D:\eye_dataset\grayscale-eye-dataset-1000-4classes"  # Folder with subfolders for classes

def load_and_preprocess_all(data_dir):
    images = []
    labels = []
    subset_path = os.path.join(data_dir)
    class_counts = {cls: 0 for cls in CLASS_NAMES}

    for idx, cls in enumerate(CLASS_NAMES):
        cls_path = os.path.join(subset_path, cls)
        if not os.path.exists(cls_path):
            print(f"Warning: class folder {cls} does not exist.")
            continue
        files = [f for f in os.listdir(cls_path) if os.path.isfile(os.path.join(cls_path, f))]
        class_counts[cls] = len(files)
        print(f"Loading {class_counts[cls]} images for class '{cls}' in  dataset")
        for file in files:
            img_path = os.path.join(cls_path, file)
            img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)  # Load single channel grayscale
            if img is None:
                print(f"Failed to load {img_path}")
                continue
            img_resized = cv2.resize(img, IMG_SIZE)
            # Convert grayscale to RGB by repeating channels
            img_rgb = np.stack([img_resized]*3, axis=-1)
            preprocessed_img = preprocess_input(img_rgb)
            images.append(preprocessed_img)
            labels.append(idx)

    images = np.array(images)
    labels = np.array(labels)

    print(f"Total images loaded for dataset: {len(images)}")
    print(f"Counts per class in dataset:", class_counts)
    return images, labels

def plot_class_distribution(labels, class_names, title="Class Distribution"):
    counter = Counter(labels)
    counts = [counter[i] for i in range(len(class_names))]
    for cls, count in zip(class_names, counts):
        print(f"{cls}: {count} images")
    plt.bar(class_names, counts, color='skyblue')
    plt.title(title)
    plt.xlabel('Emotion Class')
    plt.ylabel('Number of Images')
    plt.xticks(rotation=45)
    plt.show()

def build_model(input_shape=(*IMG_SIZE, 3), num_classes=len(CLASS_NAMES)):
    base_model = MobileNetV2(include_top=False, input_shape=input_shape, pooling='avg', weights='imagenet')

    # Freeze base model layers initially
    base_model.trainable = False

    model = models.Sequential([
        base_model,
        layers.BatchNormalization(),
        layers.Dense(128, activation='relu'),
        layers.Dropout(0.5),
        layers.Dense(num_classes, activation='softmax')
    ])

    model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
    return model

def plot_training_history(history):
    plt.figure(figsize=(12, 5))

    plt.subplot(1, 2, 1)
    plt.plot(history.history['accuracy'], label='Train Accuracy')
    plt.plot(history.history.get('val_accuracy', []), label='Validation Accuracy')
    plt.title('Training and Validation Accuracy')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy')
    plt.legend()

    plt.subplot(1, 2, 2)
    plt.plot(history.history['loss'], label='Train Loss')
    plt.plot(history.history.get('val_loss', []), label='Validation Loss')
    plt.title('Training and Validation Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()

    plt.tight_layout()
    plt.show()

def predict_eye_emotion(image_path, model):
    img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
    if img is None:
        return "Error: Unable to load image", 0.0
    img_resized = cv2.resize(img, IMG_SIZE)
    img_rgb = np.stack([img_resized]*3, axis=-1)
    preprocessed_img = preprocess_input(img_rgb)
    input_tensor = np.expand_dims(preprocessed_img, axis=0)

    predictions = model.predict(input_tensor)
    predicted_class = np.argmax(predictions[0])
    confidence = np.max(predictions[0]) * 100  # percentage

    return CLASS_NAMES[predicted_class], confidence

# Load dataset
X, y = load_and_preprocess_all(DATA_DIR)

# Plot dataset distribution
plot_class_distribution(y, CLASS_NAMES, "Full Dataset Class Distribution")

# Train-test split with stratification
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

plot_class_distribution(y_train, CLASS_NAMES, "Train Set Distribution")
plot_class_distribution(y_test, CLASS_NAMES, "Test Set Distribution")

# Compute class weights to balance training
class_weights = class_weight.compute_class_weight('balanced', classes=np.unique(y_train), y=y_train)
class_weights_dict = {i : class_weights[i] for i in range(len(class_weights))}
print("Class weights:", class_weights_dict)

# Data augmentation
train_datagen = ImageDataGenerator(rotation_range=20,
                                   shear_range=0.2,
                                   zoom_range=0.2,
                                   width_shift_range=0.1,
                                   height_shift_range=0.1,
                                   horizontal_flip=True,
                                   fill_mode='nearest')

val_datagen = ImageDataGenerator()

# Build and compile model
model = build_model()
model.summary()

# Callbacks for better training
early_stopping = callbacks.EarlyStopping(monitor='val_loss', patience=7, restore_best_weights=True)
reduce_lr = callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=3, min_lr=1e-6)

# Train model with augmentation and class weights
history = model.fit(train_datagen.flow(X_train, y_train, batch_size=BATCH_SIZE),
                    validation_data=val_datagen.flow(X_test, y_test, batch_size=BATCH_SIZE),
                    epochs=EPOCHS,
                    class_weight=class_weights_dict,
                    callbacks=[early_stopping, reduce_lr])

plot_training_history(history)

# Example prediction using the trained model
test_image = r"D:\Dataset_500_128_128\angry\img_00024.png"
emotion, confidence = predict_eye_emotion(test_image, model)
print(f"Predicted Emotion: {emotion} (Confidence: {confidence:.2f}%)")
```

```python
import cv2
import numpy as np
from tensorflow.keras.models import load_model
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input
import time
from ultralytics import YOLO

# Load your trained emotion model
emotion_model = load_model('best_emotion_model.h5')
CLASS_NAMES = ['Boring', 'Neutral', 'Stress']

# Load YOLOv8 model (it will automatically download the model if not present)
# Using the nano version for speed: 'yolov8n.pt'
model_yolo = YOLO('yolov8n.pt')  # or use 'yolov8s.pt' for better accuracy

# Initialize webcam
cap = cv2.VideoCapture(0)
cap.set(cv2.CAP_PROP_FRAME_WIDTH, 1280)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 720)

# Create window
cv2.namedWindow("Emotion Detection", cv2.WINDOW_NORMAL)

# Color coding for emotions
EMOTION_COLORS = {
    'Boring': (0, 165, 255),    # Orange
    'Neutral': (0, 255, 0),     # Green
    'Stress': (0, 0, 255)       # Red
}

# FPS calculation
prev_time = 0
curr_time = 0

# Initialize emotion state
current_emotion = "Initializing..."
current_confidence = 0.0
last_emotion_time = time.time()
last_detection_time = 0

while True:
    ret, frame = cap.read()
    if not ret:
        break
        
    # Mirror the frame for more natural interaction
    frame = cv2.flip(frame, 1)
    display_frame = frame.copy()
    
    # Calculate FPS
    curr_time = time.time()
    fps = 1 / (curr_time - prev_time) if prev_time > 0 else 0
    prev_time = curr_time
    
    # Detect persons using YOLOv8
    results = model_yolo(frame, classes=[0], verbose=False)  # class 0 is person
    
    # Initialize bbox
    bbox = None
    
    # Process results
    for result in results:
        boxes = result.boxes
        if len(boxes) > 0:
            # Get the largest person (by area)
            largest_box = None
            largest_area = 0
            for box in boxes:
                # Get coordinates
                x1, y1, x2, y2 = box.xyxy[0].tolist()
                area = (x2 - x1) * (y2 - y1)
                if area > largest_area:
                    largest_area = area
                    largest_box = (x1, y1, x2, y2)
            
            if largest_box:
                x1, y1, x2, y2 = largest_box
                w = x2 - x1
                h = y2 - y1
                bbox = (int(x1), int(y1), int(w), int(h))
                last_detection_time = time.time()
    
    if bbox:
        x, y, w, h = bbox
        
        # Draw bounding box
        cv2.rectangle(display_frame, (x, y), (x + w, y + h), (0, 255, 255), 2)
        
        # Crop the detected person
        cropped = frame[y:y+h, x:x+w]
        
        if cropped.size > 0 and w > 10 and h > 10:  # Ensure valid crop
            try:
                # Preprocess for emotion model
                resized = cv2.resize(cropped, (224, 224))
                rgb_img = cv2.cvtColor(resized, cv2.COLOR_BGR2RGB)
                preprocessed = preprocess_input(rgb_img)
                input_tensor = np.expand_dims(preprocessed, axis=0)
                
                # Make prediction
                predictions = emotion_model.predict(input_tensor, verbose=0)
                predicted_idx = np.argmax(predictions[0])
                emotion = CLASS_NAMES[predicted_idx]
                confidence = np.max(predictions[0])
                
                # Update current emotion state
                current_emotion = emotion
                current_confidence = confidence
                last_emotion_time = time.time()
                
                # Display emotion and confidence on person
                text = f"{emotion} ({confidence*100:.1f}%)"
                color = EMOTION_COLORS.get(emotion, (255, 255, 255))
                cv2.putText(display_frame, text, (x, y-10), 
                            cv2.FONT_HERSHEY_SIMPLEX, 1, color, 2)
                
                # Show the cropped region in corner
                cropped_display = cv2.resize(cropped, (200, 200))
                display_frame[10:210, 10:210] = cropped_display
                cv2.putText(display_frame, "Detection Area", (15, 225), 
                            cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255, 255, 255), 1)
                
            except Exception as e:
                # Handle small images or other processing errors
                print(f"Processing error: {str(e)}")
    
    # Always display the current emotion state
    elapsed_time = time.time() - last_emotion_time
    
    # Show "No person detected" if no detection for 1 second
    if time.time() - last_detection_time > 1.0:
        status_text = "No person detected"
        status_color = (200, 200, 200)  # Gray
    else:
        status_text = f"Current State: {current_emotion} ({current_confidence*100:.1f}%)"
        status_color = EMOTION_COLORS.get(current_emotion, (255, 255, 255))
    
    # Display status in top center
    text_size = cv2.getTextSize(status_text, cv2.FONT_HERSHEY_SIMPLEX, 1, 2)[0]
    text_x = (display_frame.shape[1] - text_size[0]) // 2
    cv2.putText(display_frame, status_text, (text_x, 50), 
                cv2.FONT_HERSHEY_SIMPLEX, 1, status_color, 2)
    
    # Display FPS in top left
    fps_text = f"FPS: {fps:.1f}"
    cv2.putText(display_frame, fps_text, (20, 50), 
                cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 255), 2)
    
    # Display YOLO version
    cv2.putText(display_frame, f"YOLOv8n", (20, 90), 
                cv2.FONT_HERSHEY_SIMPLEX, 0.7, (200, 200, 0), 2)
    
    # Display instructions in bottom left
    cv2.putText(display_frame, "Press 'q' to exit", (20, display_frame.shape[0]-30), 
                cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 2)
    
    # Show the frame
    cv2.imshow("Emotion Detection", display_frame)
    
    # Exit on 'q' key press
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Release resources
cap.release()
cv2.destroyAllWindows()
```

```python
import os
import cv2
import numpy as np
from sklearn.model_selection import train_test_split
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense, Dropout, BatchNormalization, concatenate
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, ReduceLROnPlateau
from tensorflow.keras.optimizers import Adam

# Constants
IMG_SIZE_FACE = (32, 64)  # Height, width
IMG_SIZE_HAND = (64, 64)  # Height, width
BATCH_SIZE = 32
EPOCHS = 50

# Emotion classes
eye_emotions = ['angry', 'disgust', 'fear', 'happy', 'sad', 'surprise']
hand_emotions = ['stress', 'neutral', 'boring']

def load_images_labels(dataset_path, emotions, img_size):
    images, labels = [], []
    for idx, emotion in enumerate(emotions):
        folder = os.path.join(dataset_path, emotion)
        if not os.path.exists(folder):
            print(f"Skipping missing folder: {folder}")
            continue
        for fname in os.listdir(folder):
            img_path = os.path.join(folder, fname)
            img = cv2.imread(img_path)
            if img is None:
                continue
            # Resize using (width, height) order
            img_resized = cv2.resize(img, (img_size[1], img_size[0]))
            # Convert BGR to RGB
            img_rgb = cv2.cvtColor(img_resized, cv2.COLOR_BGR2RGB)
            images.append(img_rgb)
            labels.append(idx)
    images = np.array(images, dtype=np.float32) / 255.0
    labels = np.array(labels)
    return images, labels

# Paths to datasets
face_dataset_path = r"D:\eye_dataset\150_eye_RGB_new"
hand_dataset_path = r"D:\HandMomentDataset"

# Load data
X_face, y_face = load_images_labels(face_dataset_path, eye_emotions, IMG_SIZE_FACE)
X_hand, y_hand = load_images_labels(hand_dataset_path, hand_emotions, IMG_SIZE_HAND)

# One-hot encode labels
y_face_cat = to_categorical(y_face, num_classes=len(eye_emotions))
y_hand_cat = to_categorical(y_hand, num_classes=len(hand_emotions))

# Split into train/test sets
X_face_train, X_face_test, y_face_train, y_face_test = train_test_split(
    X_face, y_face_cat, test_size=0.2, stratify=y_face, random_state=42)
X_hand_train, X_hand_test, y_hand_train, y_hand_test = train_test_split(
    X_hand, y_hand_cat, test_size=0.2, stratify=y_hand, random_state=42)

# Ensure train and test sizes match by trimming to smaller dataset size
min_train_size = min(len(X_face_train), len(X_hand_train))
X_face_train = X_face_train[:min_train_size]
y_face_train = y_face_train[:min_train_size]
X_hand_train = X_hand_train[:min_train_size]
y_hand_train = y_hand_train[:min_train_size]

min_test_size = min(len(X_face_test), len(X_hand_test))
X_face_test = X_face_test[:min_test_size]
y_face_test = y_face_test[:min_test_size]
X_hand_test = X_hand_test[:min_test_size]
y_hand_test = y_hand_test[:min_test_size]

# Build MobileNetV2 backbone
def mobilenet_backbone(input_shape, name):
    base = MobileNetV2(include_top=False, input_shape=input_shape, pooling='avg', weights='imagenet', name=name)
    base.trainable = False
    return base

# Inputs
face_input = Input(shape=(*IMG_SIZE_FACE, 3), name='face_input')  # height, width, channels
hand_input = Input(shape=(*IMG_SIZE_HAND, 3), name='hand_input')

# Feature extractors
face_base = mobilenet_backbone((*IMG_SIZE_FACE, 3), name='face_mobilenet')
hand_base = mobilenet_backbone((*IMG_SIZE_HAND, 3), name='hand_mobilenet')

face_feat = face_base(face_input)
hand_feat = hand_base(hand_input)

# Dense layers
face_dense = Dense(128, activation='relu')(face_feat)
hand_dense = Dense(128, activation='relu')(hand_feat)

# Concatenate
combined_feats = concatenate([face_dense, hand_dense])
combined_feats = BatchNormalization()(combined_feats)
combined_feats = Dropout(0.5)(combined_feats)
combined_feats = Dense(128, activation='relu')(combined_feats)

# Output heads
face_output = Dense(len(eye_emotions), activation='softmax', name='face_output')(combined_feats)
hand_output = Dense(len(hand_emotions), activation='softmax', name='hand_output')(combined_feats)

# Model
model = Model(inputs=[face_input, hand_input], outputs=[face_output, hand_output])

# Compile model
model.compile(optimizer=Adam(1e-4),
              loss={'face_output': 'categorical_crossentropy', 'hand_output': 'categorical_crossentropy'},
              metrics={'face_output': 'accuracy', 'hand_output': 'accuracy'})

model.summary()

# Data augmentation
face_datagen = ImageDataGenerator(rotation_range=20, width_shift_range=0.1,
                                 height_shift_range=0.1, zoom_range=0.2,
                                 horizontal_flip=True, fill_mode='nearest')
hand_datagen = ImageDataGenerator(rotation_range=15, width_shift_range=0.1,
                                 height_shift_range=0.1, zoom_range=0.2,
                                 horizontal_flip=True, fill_mode='nearest')

# Custom generator for multimodal batch
def combined_generator(batch_size):
    face_gen = face_datagen.flow(X_face_train, y_face_train, batch_size=batch_size, seed=42)
    hand_gen = hand_datagen.flow(X_hand_train, y_hand_train, batch_size=batch_size, seed=42)
    while True:
        f_batch, yf = next(face_gen)
        h_batch, yh = next(hand_gen)
        # Align batch sizes
        min_batch = min(f_batch.shape[0], h_batch.shape[0])
        f_batch, yf = f_batch[:min_batch], yf[:min_batch]
        h_batch, yh = h_batch[:min_batch], yh[:min_batch]
        yield (f_batch, h_batch), (yf, yh)

steps_per_epoch = min_train_size // BATCH_SIZE

# Callbacks
early_stop = EarlyStopping(
    monitor='val_face_output_accuracy',
    patience=10,
    restore_best_weights=True,
    mode='max'   # Add this line
)
reduce_lr = ReduceLROnPlateau(
    monitor='val_face_output_loss',
    factor=0.5,
    patience=4,
    verbose=1,
    mode='min'   # Add this line
)
checkpoint = ModelCheckpoint('multimodal_emotion_model.h5', monitor='val_face_output_loss', save_best_only=True)

# Train model
history = model.fit(combined_generator(BATCH_SIZE),
                    steps_per_epoch=steps_per_epoch,
                    epochs=EPOCHS,
                    validation_data=([X_face_test, X_hand_test], [y_face_test, y_hand_test]),
                    callbacks=[early_stop, reduce_lr, checkpoint])

# Evaluate
results = model.evaluate([X_face_test, X_hand_test], [y_face_test, y_hand_test])
print(f"Face test accuracy: {results[3]*100:.2f}%")
print(f"Hand test accuracy: {results[4]*100:.2f}%")
```

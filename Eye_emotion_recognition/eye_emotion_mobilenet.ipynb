```python
import os
import cv2
import numpy as np
from collections import Counter
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.utils import class_weight
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications.mobilenet_v2 import MobileNetV2, preprocess_input
from tensorflow.keras import layers, models, optimizers, callbacks

# Constants
IMG_SIZE = (224, 224)
BATCH_SIZE = 32
EPOCHS = 50
CLASS_NAMES = ['angry', 'disgust', 'fear', 'happy', 'sad', 'surprise']
DATA_DIR = r"D:\eye_dataset\150_eye dataset_gray"  # Folder with subfolders for classes

def load_and_preprocess_all(data_dir):
    images = []
    labels = []
    class_counts = {cls: 0 for cls in CLASS_NAMES}

    for idx, cls in enumerate(CLASS_NAMES):
        cls_path = os.path.join(data_dir, cls)
        if not os.path.exists(cls_path):
            print(f"Warning: class folder {cls} does not exist.")
            continue
        files = [f for f in os.listdir(cls_path) if os.path.isfile(os.path.join(cls_path, f))]
        class_counts[cls] = len(files)
        print(f"Loading {class_counts[cls]} images for class '{cls}'")
        for file in files:
            img_path = os.path.join(cls_path, file)
            img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)  # Load single channel grayscale
            if img is None:
                print(f"Failed to load {img_path}")
                continue
            img_resized = cv2.resize(img, IMG_SIZE)
            # Convert grayscale to RGB by repeating channels
            img_rgb = np.stack([img_resized]*3, axis=-1)
            preprocessed_img = preprocess_input(img_rgb)
            images.append(preprocessed_img)
            labels.append(idx)

    images = np.array(images)
    labels = np.array(labels)

    print(f"Total images loaded: {len(images)}")
    print("Counts per class:", class_counts)
    return images, labels

def plot_class_distribution(labels, class_names, title="Class Distribution"):
    counter = Counter(labels)
    counts = [counter[i] for i in range(len(class_names))]
    for cls, count in zip(class_names, counts):
        print(f"{cls}: {count} images")
    plt.bar(class_names, counts, color='skyblue')
    plt.title(title)
    plt.xlabel('Emotion Class')
    plt.ylabel('Number of Images')
    plt.xticks(rotation=45)
    plt.show()

def build_model(input_shape=(*IMG_SIZE, 3), num_classes=len(CLASS_NAMES)):
    base_model = MobileNetV2(include_top=False, input_shape=input_shape, pooling='avg', weights='imagenet')

    # Freeze base model layers initially
    base_model.trainable = False

    model = models.Sequential([
        base_model,
        layers.BatchNormalization(),
        layers.Dense(128, activation='relu'),
        layers.Dropout(0.5),
        layers.Dense(num_classes, activation='softmax')
    ])

    model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
    return model

def plot_training_history(history):
    plt.figure(figsize=(12, 5))

    plt.subplot(1, 2, 1)
    plt.plot(history.history['accuracy'], label='Train Accuracy')
    plt.plot(history.history.get('val_accuracy', []), label='Validation Accuracy')
    plt.title('Training and Validation Accuracy')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy')
    plt.legend()

    plt.subplot(1, 2, 2)
    plt.plot(history.history['loss'], label='Train Loss')
    plt.plot(history.history.get('val_loss', []), label='Validation Loss')
    plt.title('Training and Validation Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()

    plt.tight_layout()
    plt.show()

def predict_eye_emotion(image_path, model):
    img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
    if img is None:
        return "Error: Unable to load image", 0.0
    img_resized = cv2.resize(img, IMG_SIZE)
    img_rgb = np.stack([img_resized]*3, axis=-1)
    preprocessed_img = preprocess_input(img_rgb)
    input_tensor = np.expand_dims(preprocessed_img, axis=0)

    predictions = model.predict(input_tensor)
    predicted_class = np.argmax(predictions[0])
    confidence = np.max(predictions[0]) * 100  # percentage

    return CLASS_NAMES[predicted_class], confidence

# Load dataset
X, y = load_and_preprocess_all(DATA_DIR)

# Plot dataset distribution
plot_class_distribution(y, CLASS_NAMES, "Full Dataset Class Distribution")

# Train-test split with stratification
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

plot_class_distribution(y_train, CLASS_NAMES, "Train Set Distribution")
plot_class_distribution(y_test, CLASS_NAMES, "Test Set Distribution")

# Compute class weights to balance training
class_weights = class_weight.compute_class_weight('balanced', classes=np.unique(y_train), y=y_train)
class_weights_dict = {i : class_weights[i] for i in range(len(class_weights))}
print("Class weights:", class_weights_dict)

# Data augmentation
train_datagen = ImageDataGenerator(rotation_range=20,
                                   shear_range=0.2,
                                   zoom_range=0.2,
                                   width_shift_range=0.1,
                                   height_shift_range=0.1,
                                   horizontal_flip=True,
                                   fill_mode='nearest')

val_datagen = ImageDataGenerator()

# Build and compile model
model = build_model()
model.summary()

# Callbacks for better training
early_stopping = callbacks.EarlyStopping(monitor='val_loss', patience=7, restore_best_weights=True)
reduce_lr = callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=3, min_lr=1e-6)

# Train model with augmentation and class weights
history = model.fit(train_datagen.flow(X_train, y_train, batch_size=BATCH_SIZE),
                    validation_data=val_datagen.flow(X_test, y_test, batch_size=BATCH_SIZE),
                    epochs=EPOCHS,
                    class_weight=class_weights_dict,
                    callbacks=[early_stopping, reduce_lr])

plot_training_history(history)

# Example prediction using the trained model
test_image = r"D:\Dataset_500_128_128\angry\img_00024.png"
emotion, confidence = predict_eye_emotion(test_image, model)
print(f"Predicted Emotion: {emotion} (Confidence: {confidence:.2f}%)")
```

```python
import os
import cv2
import numpy as np
from tensorflow.keras import layers, models
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.callbacks import EarlyStopping
from sklearn.model_selection import train_test_split

IMG_SIZE = (224, 224)
BATCH_SIZE = 32
EPOCHS = 50
CLASS_NAMES = ['angry', 'disgust', 'fear', 'happy', 'sad', 'surprise']
DATA_DIR = r"D:\eye_dataset\150_eye dataset_gray"  # Single folder containing class subfolders

def load_data_and_split(data_dir, test_size=0.2, random_seed=42):
    images = []
    labels = []
    for idx, cls in enumerate(CLASS_NAMES):
        cls_path = os.path.join(data_dir, cls)
        files = os.listdir(cls_path)
        for file in files:
            img_path = os.path.join(cls_path, file)
            img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)  # Grayscale load
            if img is None:
                continue
            img_resized = cv2.resize(img, IMG_SIZE)
            img_normalized = img_resized.astype("float32") / 255.0  # Normalize
            img_expanded = np.expand_dims(img_normalized, axis=-1)  # Shape: (224,224,1)
            images.append(img_expanded)
            labels.append(idx)
    images = np.array(images)
    labels = np.array(labels)

    X_train, X_test, y_train, y_test = train_test_split(
        images, labels, test_size=test_size, stratify=labels, random_state=random_seed
    )
    return X_train, X_test, y_train, y_test

print("Loading and splitting data...")
X_train, X_test, y_train, y_test = load_data_and_split(DATA_DIR)

print("Train samples:", X_train.shape, "Test samples:", X_test.shape)

train_datagen = ImageDataGenerator(
    rotation_range=20,
    shear_range=0.1,
    zoom_range=0.1,
    width_shift_range=0.1,
    height_shift_range=0.1,
    horizontal_flip=True,
    fill_mode='nearest'
)
val_datagen = ImageDataGenerator()

def build_cnn_model(input_shape=(*IMG_SIZE, 1), num_classes=len(CLASS_NAMES)):
    model = models.Sequential([
        layers.Conv2D(32, (3,3), activation='relu', input_shape=input_shape),
        layers.MaxPooling2D((2,2)),
        layers.Conv2D(64, (3,3), activation='relu'),
        layers.MaxPooling2D((2,2)),
        layers.Conv2D(128, (3,3), activation='relu'),
        layers.MaxPooling2D((2,2)),
        layers.Flatten(),
        layers.Dense(128, activation='relu'),
        layers.Dropout(0.4),
        layers.Dense(num_classes, activation='softmax')
    ])
    model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
    return model

model = build_cnn_model()
model.summary()

early_stopping = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)

history = model.fit(
    train_datagen.flow(X_train, y_train, batch_size=BATCH_SIZE),
    validation_data=val_datagen.flow(X_test, y_test, batch_size=BATCH_SIZE),
    epochs=EPOCHS,
    callbacks=[early_stopping]
)
```

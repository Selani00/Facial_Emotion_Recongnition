```python
import json
import csv

# Input and output file names
jsonl_file = "dataset_dynamic_new.jsonl"
csv_file = "output.csv"

# Read .jsonl file
with open(jsonl_file, "r", encoding="utf-8") as f:
    data = [json.loads(line) for line in f]

# Get the keys (column names) from the first JSON object
keys = data[0].keys()

# Write to CSV file
with open(csv_file, "w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=keys)
    writer.writeheader()
    writer.writerows(data)

print(f"✅ Successfully converted {jsonl_file} to {csv_file}")
```

```python
import json

# Input and output file paths
input_file = "dataset_dynamic.jsonl"
output_file = "dataset_dynamic_new.jsonl"

with open(input_file, "r", encoding="utf-8") as fin, open(output_file, "w", encoding="utf-8") as fout:
    for line in fin:
        # Load each JSON object
        record = json.loads(line.strip())
        
        # Extract fields
        question = record.get("question", "")
        reasoning = record.get("reasoning", "")
        answer = record.get("answer", "")
        
        # Format new answer (reasoning + #### + original answer)
        new_answer = reasoning + "\n####\n" + answer
        
        # Build new record
        new_record = {
            "question": question,
            "answer": new_answer
        }
        
        # Write back as JSON line
        fout.write(json.dumps(new_record, ensure_ascii=False) + "\n")

print(f"✅ Converted file saved as {output_file}")
```

```python
import json

# Input and output file paths
input_file = "dataset_dynamic.jsonl"
output_file = "dataset_dynamic_1.jsonl"

with open(input_file, "r", encoding="utf-8") as infile, open(output_file, "w", encoding="utf-8") as outfile:
    for line in infile:
        if not line.strip():
            continue  # skip empty lines

        # Parse the JSON line
        data = json.loads(line)

        # Convert everything back to string, replace sequences, then re-parse JSON
        fixed_line = json.dumps(data)
        # Replace any occurrence of 8 backslashes with 2 backslashes
        fixed_line = fixed_line.replace("\\\\\\\\", "\\")
        
        # Write cleaned JSON line
        outfile.write(fixed_line + "\n")

print(f"✅ Conversion complete. Cleaned file saved as {output_file}")
```
